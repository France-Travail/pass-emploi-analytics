```python
import pandas as pd, numpy as np
```

```python
df_ml = pd.read_excel('data_beneficiaire/CEJ_ML_stk.xlsx', sheet_name="Département", skiprows=1)
df_pe = pd.read_excel('data_beneficiaire/CEJ_PE_stk.xlsx', sheet_name="Département", skiprows=1)
```

```python
df_ml['structure'] = 'MILO'
df_pe['structure'] = 'POLE_EMPLOI'
```

```python
df = pd.concat([df_ml, df_pe])
```

```python
df_pivoted = df.melt(
    id_vars = ['structure', 'Code_Département'],
    var_name='mois',
    value_name='nb_jeunes_cej',
    value_vars=[col for col in df.columns if col not in ["Code_Région", "Libellé_Région","Code_Département", "Libellé_Département"]]
)
```

```python
df_pivoted['mois'] = df_pivoted['mois'] + '-01'
```

```python
df_pivoted['nb_jeunes_cej'] = df_pivoted['nb_jeunes_cej'].replace({'-': 0})
```

```python
df_pivoted.to_csv("stock jeunes par departement.csv", index=None)
```
